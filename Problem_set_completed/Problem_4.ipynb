```python
import pandas as pd

df = pd.read_csv("../data/heart_failure_clinical_records_dataset.csv")
```

```python
#part a
ax=df["ejection_fraction"].plot.kde()
ax.set_title("KDE of Ejection Fraction")
```

```python
#part b
import matplotlib.pyplot as plt

for i in range(15):
    sample = df.sample(frac=1.0, replace=True)
    sample["ejection_fraction"].plot.kde()
plt.title("Resampled Datasets")
plt.show()
```

Part c: 
- The plot appears mostly reliable around 20-50 and the most variation occurs at 35-40 and 55-65 where the resampled curves spread apart more.
